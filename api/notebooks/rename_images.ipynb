# Rename Images
This notebook renames all images in the `raw_data/images` folder from `(1).jpeg`, `(2).jpeg`, ... to `img_00.jpeg`, `img_01.jpeg`, ...

```python
import os
import re
from pathlib import Path

# Path to the images folder
images_dir = Path("raw_data/images")

# Get all image files and sort them numerically
image_files = sorted(
    images_dir.iterdir(),
    key=lambda f: int(re.search(r'\((\d+)\)', f.stem).group(1)) if re.search(r'\((\d+)\)', f.stem) else 0
)

print(f"Found {len(image_files)} images to rename\n")

# Rename each file
for idx, file_path in enumerate(image_files):
    ext = file_path.suffix  # e.g. .jpeg
    new_name = f"img_{idx:02d}{ext}"
    new_path = images_dir / new_name
    
    print(f"{file_path.name}  ->  {new_name}")
    os.rename(file_path, new_path)

print(f"\nDone! Renamed {len(image_files)} images.")
```

```python
# Verify the results
renamed_files = sorted(images_dir.iterdir())
for f in renamed_files:
    print(f.name)
```
